I used a wine review dataset and focused on price, review points, and country. I wanted to look at the 5 largest wine producers in the world, simply because I assume those have the most data and are would be most available in stores. 

The overall statistics show that wine prices are much more spread out than wine scores. The mean price was about 35.36, but the maximum price was 3300, which creates a very large range. On the other hand, the average points value was about 88.45, and points only ranged from 80 to 100. The scatter plot made this pattern more apparent to me. Most wines were concentrated at lower prices, but their point values were still high regardless, this tells me that for the most part even if wine prices are high or low that doesn't necesarily mean that they are going to be lower quality or less enjoyable. 
When I looked at specifically the  country categories, French wines had the highest average price, then Italian wines and the American wines, while Spanish and Argentine wines had lower average prices. And yet once again, the average review scores did not differ as much as prices did. France had the highest mean points, but Italy and the United States were extremely close, and even Spain and Argentina remained within a reasonably narrow range. SO, for this tells me that when we are looking for wines it would be wise to focus specifically in Spanish and Argentine wines, because unfortunately we are still college students and we like to still be affordable not necessaarily the best of the best, that's for after graduation.
