```python
temps_louisville={'january':[43,28],'february':[48,30],'march':[58,38],'april':[69,48],'may':[78,58],'june':[86,67],'july':[89,88],'august':[88,69],'september':[82,62],'october':[71,50],'november':[59,40],'december':[48,33]}
month=input('Enter a month for the average high and low temperatures in Louisville:(write quit to stop searching) ')
while month.lower() !='quit':
    month_lower=month.lower()
    if month_lower in temps_louisville:
        high=temps_louisville[month_lower][0]
        low=temps_louisville[month_lower][1]
        print('The average high temperature for '+ month.title() + ' is ' + str(high) + ' and the average low is ' + str(low) + '.')
    else: 
        print('That is not a valid month')
    month=input('Enter a month for the average high and low temperatures in Louisville:(write quit to stop searching)')
```

```python
inventory ={'hammer':[15, 12.49], 'wrench':[20, 7.99], 'screwdriver': [40, 3.99], 'nailgun':[10, 36.59], 'padlock':[25, 5.99], 'roll of duct tape':[30, 4.95]}
choice=0
while choice !=5:
    print('Menu')
    print('(1) Search for an item')
    print('(2) SHow items for below a budget')
    print('(3) Show total quantity') 
    print('(4) Show total inventory value')
    print('(5) Quit')

    choice=int(input('Enter the number for your choice'))
    if choice == 1:
        item_name=input('Enter an item name: ').lower()
        if item_name in inventory:
            quantity=inventory[item_name][0]
            price=inventory[item_name][1]
            print('there are ' + str(quantity) + ' ' + item_name + '(s) available at $' + str(price) + 'each.')
        else: 
            print('that item is not available.')
    elif choice == 2:
        budget=float(input('Enter your budget amount: '))
        print('Items below your budget:')
        for item in inventory:
            if inventory[item][1]< budget:
                print(item + ': $' + str(inventory[item][1])) 
    elif choice == 3:
        total_quantity=0
        for item in inventory: 
            total_quantity= total_quantity + inventory[item][0]
        print('the total quantity of item is' + str(total_quantity) + '.')
    elif choice == 4:
        total_price=0 
        for item in inventory:
            quantity=inventory[item][0]
            price=inventory[item][1]
            total_price=total_price + (quantity*price)
        print('the price total of all the inventory is ' + str(round(total_price, 2)) + '$.')
    elif choice == 5: 
        print('Quitting')
    else: 
        print('Try again')

```
